```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

# 데이터 로드
df = pd.read_csv('final_evaluation_summary.csv')
img_dir = '/home/ai3/AD-content/test_data'

# 시각화 함수
def display_results(df, img_dir, num_samples=5):
    fig, axes = plt.subplots(num_samples, 2, figsize=(10, 3 * num_samples))
    
    for i in range(num_samples):
        row = df.iloc[i]
        img_path = os.path.join(img_dir, row['File'])
        
        # 이미지 출력
        axes[i, 0].imshow(mpimg.imread(img_path))
        axes[i, 0].axis('off')
        axes[i, 0].set_title(row['File'])
        
        # 지표 텍스트 출력
        metrics_text = f"PSNR: {row['PSNR']:.2f}\nSSIM: {row['SSIM']:.2f}\nLPIPS: {row['LPIPS']:.2f}\nCLIP: {row['CLIP_Score']:.2f}"
        axes[i, 1].text(0.1, 0.5, metrics_text, fontsize=12, va='center')
        axes[i, 1].axis('off')
        
    plt.tight_layout()
    plt.show()

# 실행
display_results(df, img_dir)
```
